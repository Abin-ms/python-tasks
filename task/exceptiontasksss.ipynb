```python
# 1.write a program to divide two numbers using try and - except.
try:
    num1 = 10
    num2 = 0
    result = num1 / num2
    print(result)
except:
    print("Numbers cannot be divided")
```

```python
# handle value Error while converting a string into an integer.
try:
    strr = int(input("Enter the string :"))
    print(strr)
except ValueError:
    print("String cannot be converted to integer")
```

```python
# handle zerrodivisionerror and valueError using seperate except blocks
try:
    num1 = int(input("Enter the first number : "))
    num2 = int(input("Enter the second number : "))
    result = num1/num2
    print(result)
except ZeroDivisionError:
    print("Number cannot be divided by 0")
except ValueError:
    print("String cannot be accepted")
```

```python
# Write a program using try, except and else.
try:
    num1 = 10
    num2 = 2
    result = num1 / num2

except Exception:
    print("Exception occured handle it properly")
else:
    print("result is : ",result)
```

```python
# Write a program using try, except and finally.
try:
    num1 = 10
    num2 = 2
    result = num1 / num2
    print("result is : ",result)

except Exception:
    print("Exception occured handle it properly")
finally:
    print("Executed succesfully")
```

```python
# Handle IndexError while accessing a list element.
try:
    listt = [1,2,3,4]
    print(listt[6])
except IndexError:
    print("index is not found...!!")
```

```python
# Handle KeyError while accessing a dictionary.
try:
    diii = { "a":"adad","b":"adadad"}
    print(diii["a"])
except KeyError:
    print("Key not is not founded dictionary..!!!")
```

```python
# Write a program to handle multiple exceptions using a single except block

try:
    num1 = int(input("Enter the first number : "))
    num2 = int(input("Enter the second number : "))
    result = num1/num2
    print(result)
except (ZeroDivisionError,ValueError):
    print("ZeroDivisionError or ValueError occured...!!!")

# except ValueError:
#     print("String cannot be accepted")
```

```python
# Create a nested try-except program.
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    try:
        result = num1 / num2
        print("Result:", result)

    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")

except ValueError:
    print("Error: Please enter valid integers!")
```
