```python
# program to find sum and product of digits in a number.
num = int(input("hahah"))
temp = 0
num_reverse = 0
sum = 0
product = 1
while( num!=0):
    temp = num % 10
    num_reverse = num_reverse * 10 + temp
    sum = sum + temp
    product = product * temp
    num = num // 10
print(num_reverse)
print(sum)
print(product)

```

```python
# take a number ( let the number be a two digit number).
# reverse the number.
# check if it is a magic number or not ( if digit1 + digit2 = 1 : it is a magic number or else it is not a magic number )
num = int(input("Enter a number"))
temp = 0
num_reverse = 0
sum = 0
while( num != 0):
    temp = num % 10
    sum = sum + temp
    num_reverse = num_reverse * 10 + temp
    num = num // 10

if(sum % 10 == 1):
    print("It is a magic number")
else:
    print("It is not a magic number")
```
