```python
# print numbers from 1 to n.

limit = int(input("Enter the limit-n"))

i = 1
while(i <= limit):
    print(i)
    i+=1
```

```python
#  find the sum of numbers from 1 to n.

num = int(input("Enter the limit : "))

sum = 0
i = 1
while(i <= num):
    sum = sum + i
    i+=1
print(sum)
```

```python
# find the sum of even numbers from 1 to n.

num = int(input("Enter "))

sum = 0
i = 1
while(i <= num):
    if( i % 2 == 0):
        sum = sum + i
    i+=1
print(sum)
```

```python
#  find the sum of odd numbers from 1 to n.
num = int(input("Enter "))

sum = 0
i = 1
while(i <= num):
    if( i % 2 != 0):
        sum = sum + i
    i+=1
print(sum)
```

```python
# Count how many numbers are present from 1 to n

num = int(input("Enter the range"))

count = 0

i = 1
while(i <= num):
    count+=1
    i+=1

print(count)
```
