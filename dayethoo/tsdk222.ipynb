```python
# create a list and find the number of palindrome in the list Q(1)
list1 = [1112, 121, 212, 34, 56, 76, 77, 11]

count = 0

for num in list1:
    temp = num
    rev = 0

    while temp != 0:
        digit = temp % 10
        rev = rev * 10 + digit
        temp = temp // 10

    if rev == num:
        count += 1

print("The number of palindromes in the list:", count)
```

```python
# count frequency of a word without split() 
word = input("Enter a word: ")
visited = ""

for ch in word:
    if ch not in visited:
        count = 0

        for x in word:
            if ch == x:
                count += 1

        print(ch, "=", count)
        visited += ch
```

```python
# Q6

# count frequency of a word without split() 
word = input("Enter a word: ")
visited = ""

for ch in word:
    if ch not in visited:
        count = 0

        for x in word:
            if ch == x:
                count += 1

        if count == 1:
            print(ch,"is the first non repeating character...")
            break
 
        
        visited += ch
```
