```python
# create a list and find the number of palindrome in the list Q(1)
list1 = [1112, 121, 212, 34, 56, 76, 77, 11]

count = 0

for num in list1:
    temp = num
    rev = 0

    while temp != 0:
        digit = temp % 10
        rev = rev * 10 + digit
        temp = temp // 10

    if rev == num:
        count += 1

print("The number of palindromes in the list:", count)
```

```python
# using string.(Q1)

list1 = ["aae","aaa","aba","acca","ahhaa"]
count = 0
for name in list1:
    temp = name[::-1]
    if(temp == name):
        count+=1
print("Number of palindrome in the list :",count)
```

```python
# Q2

list1 = [111, 121, 212, 343, 565, 767, 77, 11]
largest = 0
smallest = list1[0]
for li in list1:
    if(li > largest):
        largest = li
    if(li < li):
        smallest = li

print(largest)
print(smallest)
```
