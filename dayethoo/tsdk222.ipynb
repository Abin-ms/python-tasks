```python
# list tutorial

lisstt = [20,10,30,80,50,60,90,40,70]
lisstt.reverse()
print(lisstt)
print(len(lisstt))
print(lisstt)

print(max(lisstt))
print( min(lisstt) )
print(sum( lisstt))
print( sorted(lisstt))
print( any(lisstt) )
print(all(lisstt))
print(lisstt[0])
print( sorted(lisstt))
lisstt.reverse()
print(lisstt)
```

```python
lisstt = [20,10,30]

lisstt.append(40)
print(lisstt)
lisstt.reverse()
print(lisstt)
```

```python
lisss = [10,20,30,40]

print(lisss)
lisss.append(60)
print(lisss)

lisss.insert( 10 , 200)
print(lisss)
print(lisss[5])
```

```python
# .extend() use case.

list1 = [1,2,3,4,5]
print(list1)
list1.extend([3,2,3,3,4])
print(list1)
# nested list.
list1.append([12,33,55,555])
print(list1)
```
