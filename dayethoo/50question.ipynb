```python
num = int(input("Enter a number: "))

n = num + 1
i = 1

while i * i <= n:
    if i * i == n:
        print(num, "is a Sunny Number")
        break
    i += 1
else:
    print(num, "is not a Sunny Number")
```

```python
import math

num = int(input("Enter a number: "))

n = num + 1
root = math.isqrt(n)

if root * root == n:
    print(num, "is a Sunny Number")
else:
    print(num, "is not a Sunny Number")
```

```python
# number is spy or not.
num = int(input("Enter a number"))
n = num
sum = 0
product = 1
while num != 0:
    digitt = num % 10

    sum = sum + digitt
    product = product * digitt

    num = num // 10

if(sum == product):
    print(n,"is a spy number")
else:
    print(n,"is not a spy number")
```
