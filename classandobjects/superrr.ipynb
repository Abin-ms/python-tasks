```python
class School():
    def __init__(self,loc,s_name):
        self.loc = loc
        self.s_name = s_name
        print("Inside parent : ",s_name)
        print("Inside parent : ",loc)
        

class child(School):
    def __init__(self,loc,s_name,stu_name,stu_grade):
        print("inside child : ",s_name)
        print("inside child : ",loc)

        super().__init__(loc,s_name)
        self.stu_name = stu_name
        self.stu_grade = stu_grade
        print("inside child : ",self.s_name)
        print("inside child : ",self.loc)
        print(s_name)
        print(loc)
        print(self.stu_name)
        print(self.stu_grade)
        print(stu_name)
        print(stu_grade)

c = child("Alappuzha","SNMHSS","Abin","B")
        
        
```

```python
class parent():
    def __init__(self,p_name):
        self.p_name = p_name
        print(p_name)
class child(parent):
    def __init__(self, p_name,c1_name):
        super().__init__(p_name)
        self.c1_name = c1_name
class child2(child):
    def __init__(self, p_name, c1_name,c2_name):
        super().__init__(p_name, c1_name)
        self.c2_name = c2_name
        print("Parent name : ",p_name)
        print("Child1 name : ",c1_name) 
        print("Child2 name : ",c2_name)
c = child2("P1","ch1","ch2")

        
```

```python
class Parent():
    def __init__(self,p_name):
        self.p_name = p_name
class Parent2():
    def __init__(self,age,work):
        self.age = age
        self.work = work
class child(Parent,Parent2):
    def __init__(self,c_name):
        Parent.__init__(self,"haha")
        Parent2.__init__(self,20,"ajaj")
        self.c_name = c_name

c = child("Abin")
print(c.p_name)
print(c.age)
print(c.work)
print(c.c_name)
```
