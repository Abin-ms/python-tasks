```python
# Q16. Change Student Details
# Create a class Student with a constructor that accepts name and age.
# Create a method update_details() that changes the student's name and age. Create another method to display the updated details.
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def update_details(self):
        self.name = input("Enter name")
        self.age = int(input("Enter new age"))
    def display(self):
        print(self.name)
        print(self.age)

s1 = Student("Abin",22)

print("Original details")
s1.display()

print("Updating details...")
s1.update_details()

print("Updated details..")
s1.display()

```

```python
#  Q17. Two Objects, Different Values
#  Create a class Person with a parameterized constructor that accepts name and city. Create two objects with different values.
#  Create a method to display each person's details.
class Person:
    def __init__(self,name,city):
        self.name = name
        self.city = city
    def display(self):
        print("Name : ",self.name)
        print("City : ",self.city)
s1 = Person("Abin","Alappuzha")
s2 = Person("ajakdj","kunnumkulam")
s1.display()
s2.display()
```

```python
#  Q18. Constructor and Multiple Operations
# Create a class BankAccount with a constructor that accepts name and balance. Create methods to:
# Deposit money.
# Withdraw money, provided the balance is sufficient.
# Display the current balance.

class BankAccount:
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance
    def deposit(self):
        amount = int(input("Enter the amount to deposite:"))
        self.balance = self.balance + amount
    def withdraw(self):
        amount = int(input("Enter the amount to withdraw : "))
        if(amount > self.balance):
            print("insufficient balance!!!!!")
        else:
            self.balance = self.balance - amount
    def balance(self):
        print("Available balance is : ",self.balance)

```
