```python
# Take a year and determine whether it is a leap year using the correct 400/100/4 rules.
year = int(input("Enter a year: "))

if year % 400 == 0:
    print(year, "is a leap year")
elif year % 100 == 0:
    print(year, "is not a leap year")
elif year % 4 == 0:
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

```python
# Take three sides and determine whether they can form a valid triangle. If valid, determine Equilateral, Isosceles, or Scalene.

side1 = int(input("Enter side 1"))
side2 = int(input("Enter side 2"))
side3 = int(input("Enter side 3"))
if( (side1 + side2 > side3) & (side2 + side3 > side1) & (side3 + side1 > side2)):
    if(side1 == side2 == side3):
      print("the triangle is equilateral")
    elif((side1 == side2) | (side2 == side3) | (side1 == side3)):
      print("The triangle is isoscless")
    else:
      print("The triangle is scalen")
else:
   print("Invalid!!!1")
```
