```python
# call a function with argument '0' , and print 0 2 4 6 8 10
def sample(i):
    if(i <= 10):
        print(i)
        # i+=2
        sample(i + 2)
sample(0)
```

```python
# create a function show() pass  the arguments from zero , if it reaches 5 i want to break the recurssion

def show(i):
    if(i <= 5):
        i+=1
    print("Reached 5 breaking  the recurssion")

show(0)
        
```
