```python
# . Create a list of fruits and print all elements using an iterator and for loop.
fruu = ["apple","orange","Mango","Grape"]
iterator = iter(fruu)
for i in iterator:
    print(i)
```

```python
# Create a generator function that generates numbers from 10 to 50 with a difference of 10.
def generatorrr():
    for i in range(10,51,10):
        yield i

for  ii in generatorrr():
    print(ii)
```

```python
# Create a generator function that generates the multiplication table of 5.
def mul5():
    for i in range(5,51,5):
        yield i
n = 1
for i in mul5():
    print(n , "   X  5 = ", i  )
    n+=1
```

```python
# Create a generator function that generates squares of numbers from 1 to 5.
def squ():
    for i in range(1,6):
        yield i * i
for i in squ():
    print(i)
```

```python
# Create a generator function that generates numbers divisible by 5 between 1 and 50
def nuu():
    for i in range(1,51):
        if i % 5 == 0:
            yield i
for i in nuu():
    print(i)
```

```python
#  Create a generator function that generates the characters of "PYTHON" one by one.
def chh():
    st = "python"
    for i in st:
        yield i 
for i in chh():
    print(i)
```

```python
# Create a generator function that generates the first 10 natural numbers and calculate their sum. 
def nat():
    for i in range(1,11):
        yield i
sum = 0
for i in nat():
    print(i)
    sum = sum + i

print(sum)
```
